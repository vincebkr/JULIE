### Ce notebook effectue un nettoyage du dataset _Speed+Dating+Data+Key.csv_ avec **pandas**

#### Chargement des données

```python
import pandas as pd

# Chargement des données
dataset = pd.read_csv('../data/raw/Speed+Dating+Data.csv')

display(dataset.head())

print("Nombre de lignes et de colonnes dans le dataset :", dataset.shape)
```

#### Suppression des colonnes avec plus de 75% de valeurs manquantes

```python
mean = 0.75
cols_to_drop = dataset.columns[dataset.isna().mean() > mean]
dataset.drop(columns=cols_to_drop, inplace=True)
print("Nombre de colonnes après suppression des colonnes avec plus de 75% de valeurs manquantes :", dataset.shape[1])
```

#### Harmonisation des échelles de notation
Certaines colonnes de notation vont de 0–100, d'autres de 0–10. On ramène tout à 0–10

```python
prefixes = ('attr', 'sinc', 'intel', 'fun', 'amb', 'shar')
rating_cols = [c for c in dataset.columns if c.startswith(prefixes)]
for col in rating_cols:
    max_val = dataset[col].dropna().astype(float).max()
    if max_val > 11:
        dataset[col] = dataset[col] / 10.0
        print(f"Rescaling {col} (0-100 ➔ 0-10)")
```

### Sauvegarde du dataset nettoyé

```python
output = '../data/raw/speed_dating_data_clean.csv'
dataset.to_csv(output, index=False)
print(f"Fichier sauvegardé : {output}")
```
